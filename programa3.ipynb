```python
def leeer_datos():
    nombres=[]
    sueldos=[]
    for i in range(5):
        nombre=input("escribe el nombre del trabajador"+str(i+1)+":")
        sueldo=float(input("escribe el sueldo de"+nombre+":"))
        nombres.append(nombre)
        sueldos.append(sueldo)
    return nombres,sueldos

def obtener_nomina_y_mayor(nombres,sueldos):
    total=0
    mayor_sueldo=0
    nombre_mayor=nombres[0]
    for i in range(len(sueldos)):
        total=total+sueldos[i]
        if sueldos[i]>mayor_sueldo:
            nombre_mayor=nombres[i]
            mayor_sueldo=sueldos[i]
    return total,nombre_mayor,mayor_sueldo

def imprimir_resultados(total,nombre_mayor,mayor_sueldo):
  print("RESULTADOS")
  print("la nomina total es:",total)
  print("la persona que mas gana es:",nombre_mayor,"con un sueldo de:",mayor_sueldo)
  print("fin del programa ")

def main():
    nombres,sueldos=leeer_datos()
    total,nombre_mayor,mayor_sueldo=obtener_nomina_y_mayor(nombres,sueldos)
    imprimir_resultados(total,nombre_mayor,mayor_sueldo)

main()
```
